## Tarea 5: Aplicación de la ecuación de Breguet

### Datos preliminares

- Del software X-Plane, se determinan los pesos $W_{fuel}, W_{EMPTY}$.
- De la tarea anterior se conoce la combinación óptima $q_p, \frac{c_L}{c_D}$, con una velocidad de $94 kn$.
- Del enunciado se conoce el rango de vuelo de la ruta.

```python
W_fuel  = 226                       # [lb]
W_EMPTY = 1721                      # [lb]
q_p     = 0.51 * 4.45 / 746 / 3600  # [m^-1]
eta_h   = 0.86
ef_aer  = 11.45
R       = (24+17+26) * 1852         # [m]
```

### Cálculos de combustible

Se solicita que al sobrevolar el destino haya un 20% del combustible, es decir: $$ W_{vuelo} = W_{FULL} + 0.2 \bullet W_{fuel} $$

Además, despejando de la ecuación de Breguet: $$ W_{FULL} = W_{EMPTY} e^{R \frac{q_p}{\eta_h} \frac{c_L}{c_D}} $$

```python
from numpy import exp

W_FULL = W_EMPTY * exp(R*q_p/eta_h / ef_aer)

W_vuelo = W_FULL + 0.2 * W_fuel

print(W_FULL, W_vuelo)
```
